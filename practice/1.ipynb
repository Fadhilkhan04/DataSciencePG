```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Create sample CVE dataset
df = pd.DataFrame({
    'CVE_ID': ['CVE-101', 'CVE-102', 'CVE-103', 'CVE-104', 'CVE-105',
               'CVE-106', 'CVE-107', 'CVE-108', 'CVE-109', 'CVE-110'],
    'Severity': ['Critical', 'High', 'Medium', 'Low', 'High',
                 'Critical', 'Medium', 'High', 'Low', 'Critical'],
    'CVSS_Score': [9.8, 7.5, 5.4, 2.1, 8.2, 9.5, 4.8, 7.8, 1.9, 9.1],
    'Year': [2022, 2023, 2022, 2021, 2023, 2024, 2021, 2024, 2022, 2024]
})

# Display first five records
print(df.head())

# Dataset information
print("\nDataset Information:")
df.info()

# Statistical summary
print("\nStatistical Summary:")
print(df.describe())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Count vulnerabilities by severity
print("\nSeverity Count:")
print(df['Severity'].value_counts())

# 1. Bar chart: Severity distribution
df['Severity'].value_counts().reindex(
    ['Critical', 'High', 'Medium', 'Low'], fill_value=0
).plot(kind='bar')

plt.title("Vulnerabilities by Severity")
plt.xlabel("Severity")
plt.ylabel("Number of Vulnerabilities")
plt.show()

# 2. Histogram: CVSS score distribution
plt.hist(df['CVSS_Score'], bins=5, edgecolor='black')
plt.title("Distribution of CVSS Scores")
plt.xlabel("CVSS Score")
plt.ylabel("Frequency")
plt.show()

# 3. Bar chart: Vulnerabilities by year
df['Year'].value_counts().sort_index().plot(kind='bar')

plt.title("Vulnerabilities by Year")
plt.xlabel("Year")
plt.ylabel("Number of Vulnerabilities")
plt.show()

# Average CVSS score by severity
print("\nAverage CVSS Score by Severity:")
print(df.groupby('Severity')['CVSS_Score'].mean())
```
