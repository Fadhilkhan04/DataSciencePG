```python
#create and manipulate numpy arrays
import numpy as np
a = np.array([1,2,3,4,5])
print("array :",a)
print("addition :",a+2)
print("multiplication :",a*2)
```

```python
#slicing and indexing
import numpy as np
a = np.array([1,2,3,4,5])
print("indexing :",a[2])
print("slicing:",a[1:3])
print("fancy indexing:",a[[0,2,4]])



#aggregate functions
print("sum :",np.sum(a))
print("mean :",np.mean(a))
print("min:",np.min(a))
print("max:",np.max(a))
```

```python
#boolean logic and masking
a = np.array([10,20,30,40,50])
print(a[a<26])
```

```python
# structured array

students = np.array([('ali',20),('john',21)])
print(students)
```

```python
#pandassssssss
#creating dataframes from dictionary , list and csv file

import pandas as pd

data1 = {'name':['ali','john','doe'],'age':[20,21,22]}  #dictionary

data2 = [['ali',20],['john',21],['doe',22]]   #list

df1 = pd.DataFrame(data1)

df2 = pd.DataFrame(data2,columns=['name','age'])

print(df1)
print(df2)

df = pd.read_csv('data.csv')
print(df)

print(df1.head())
print(df1.info())
print(df1.describe())
```

```python
df = pd.DataFrame({'name':['ali','john','doe'],'age':[20,21,22]})
print(df['name'])
print(df['age'])
print(df[df['age']>20]) # only prints the true values 
df.loc[0,'age'] = 25
print(df)
print(df.sort_values('age'))
print(df['age'].min())
print(df['age'].mean())
print(df['age'].max())



```

```python
df = pd.DataFrame({'name':['ali','john','doe'],'age':[20,np.nan,22]})
print(df.isnull())
df = df.fillna(0)
print(df.isnull())
```
