```python
import pandas as pd
import plotly.express as px
```

```python
df = px.data.gapminder()
```

```python
print("first five rows:")
print(df.head())
```

```python
print(df.shape)
```

```python
print(df.info())
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
fig = px.choropleth(df,locations="iso_alpha",color="pop",hover_name="country",hover_data={"iso_alpha":False,"pop":True,"lifeExp":True,"gdpPercap":True,"country":False},animation_frame="year",color_continuous_scale="Viridis",projection="natural earth",labels={"pop":"population","lifeExp":"Life Expectancy","gdpPercap":"GDP per capita"},title="World Population Analysis")
```

```python
fig.update_layout(title={"text":"World Population Analysis","x":0.5},geo=dict(showframe = False,showcoastlines = True))
```

```python
fig.show()
```
